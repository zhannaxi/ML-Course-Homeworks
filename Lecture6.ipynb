1. Series — негізгі тапсырмалар
temperatures деген Series жасаңыз. Ішінде қала температуралары болсын:
Алматы: 5
Астана: –3
Шымкент: 7
 Осы Series-ті экранға шығарыңыз.
Алматының температурасын индекстің көмегімен алыңыз.
Температурасы 0-ден жоғары қалаларды сүзіңіз.

```python
import pandas as pd
weather = {
    "Sity": ["Almaty", "Astana", "Shymkent"],
}
weather = pd.DataFrame(weather)

temperatures = pd.Series([5, -3, 7])
weather["Temperature"] = temperatures
weather
```

```python
print(weather.loc[0, 'Temperature'])
```

```python
filter = weather[weather["Temperature"] > 0]
filter
```

Тапсырмалар:
Бағасы 1000 теңгеден қымбат дәрілерді шығарыңыз.


Бағасы 600-800 аралығындағы дәрілерді таңдаңыз.


Қордағы саны (amount) 5-тен аз дәрілерді шығарыңыз.


```python
drugs_df = {
    "drug": ["A", "B", "C", "D"],
    "price": [1200, 800, 600, 2000],
    "amount": [5, 10, 3, 8]
}
drugs_df = pd.DataFrame(drugs_df)

filter1 = drugs_df[drugs_df["price"] > 1000]
filter2 = drugs_df[(drugs_df["price"] <= 800) & (drugs_df["price"] >= 600)]
filter3 = drugs_df[drugs_df["amount"] < 5]
print(f'Бағасы 1000 тг-ден жоғары дәрілер:\n\n{filter1}\n')
print(f'Бағасы 600-800 аралығындағы дәрілер:\n\n{filter2}\n')
print(f'Қордағы саны (amount) 5-тен аз дәрілер:\n\n{filter3}\n')
```

Тапсырмалар:
Әр бөлім бойынша орташа жалақыны есептеңіз.
Әр бөлімдегі қызметкерлер санын шығарыңыз.
Ең жоғары жалақысы бар бөлімді анықтаңыз.


```python
employees = {
    "name": ["Ayan", "Dana", "Miras", "Aida", "Timur"],
    "department": ["IT", "HR", "IT", "HR", "Sales"],
    "salary": [400000, 300000, 450000, 320000, 280000]
}
employees = pd.DataFrame(employees)

grouped_mean = employees.groupby('department')['salary'].mean()
print(f'Әр бөлім бойынша орташа жалақы:\n\n{grouped_mean}\n')

grouped_sum = employees.groupby('department')['department'].count()
print(f'Әр бөлімдегі қызметкерлер саны: \n\n{grouped_sum}\n')

grouped_max = employees.groupby('department')['salary'].max()
print(f'Әр бөлім бойынша ең жоғары жалақы: \n\n{grouped_max}\n')

max_salary = grouped_max.max()
print(f'Ең жоғары жалақысы бар бөлім: {grouped_max.idxmax()} ({max_salary} KZT)')
```

6. Агрегаттау (sum, mean, max, min)
Берілген кестеге сүйеніп:
Әр бөлім үшін жалақының ең кішісі мен ең жоғарғысын табыңыз.
Әр топтағы жалақының суммасын есептеңіз.


```python
grouped_min = employees.groupby('department')['salary'].min()
print(f'Әр бөлім бойынша ең кіші жалақы: \n{grouped_min}\n')

min_salary = grouped_min.min()
print(f'Ең кіші жалақысы бар бөлім: {grouped_min.idxmin()} ({min_salary} KZT)\n')

group_salary_sum = employees.groupby('department')['salary'].sum()
print(f'Әр топтағы жалақының суммасы:\n{group_salary_sum}')
```

Берілген екі кесте:

Тапсырмалар:
Екі кестені id бойынша біріктіріңіз (merge).
Индексті id етіп қойып, join() қолданып біріктіріңіз.
concat() қолданып students кестесін екі рет артынан қосыңыз.
concat(axis=1) қолданып students және scores кестесін бүйірінен біріктіріңіз.


```python
students2 = pd.DataFrame({
    "id": [1, 2, 3, 4],
    "name": ["Aigerim", "Dias", "Ayan", "Sara"],
})
scores = pd.DataFrame({
    "id": [1, 2, 3, 5],
    "math": [88, 75, 95, 92],
    "physics": [90, 80, 92, 87]
})
merged1 = pd.merge(students2, scores, on = "id")
print(f'Екі кестені id бойынша біріктіру:\n{merged1}\n')

merged2 = pd.merge(students2, scores, on = "id", how = "right")
print(f'Екі кестені merge(right) бойынша біріктіру:\n{merged2}\n')

merged3 = pd.merge(students2, scores, on = "id", how = "left")
print(f'Екі кестені merge(left) бойынша біріктіру:\n{merged3}\n')

merged4 = pd.merge(students2, scores, on = "id", how = "outer")
print(f'Екі кестені merge(outer) бойынша біріктіру:\n{merged4}\n')
```

```python
join_left = students2.set_index("id")
join_right = scores.set_index("id")

result_left = join_left.join(join_right)
print("Left Join:\n", result_left)

result_inner = join_left.join(join_right, how='inner')
print("\nInner Join:\n", result_inner)
```

```python
concat1 = pd.concat([students2, students2])
print(f'concat() қолданып students кестесін екі рет артынан қосу:\n{concat1}')
```

```python
concat2 = pd.concat([students2, students2], axis=1)
print(f'concat(axis=1) қолданып students және scores кестесін бүйірінен біріктір:\n{concat2}')
```

Әр студенттің орташа балын есептеп, жаңа avg бағанын қосыңыз.
Орташа балы 85-тен жоғары студенттерді таңдаңыз.


```python
df3 = pd.DataFrame({
    "student": ["Aigerim", "Dias", "Ayan"],
    "math": [88, 75, 95],
    "physics": [90, 80, 92],
    "history": [78, 85, 90]
})
print(f'DataFrame бойынша әр пәннен жалпы статистика:\n{df3.describe()}\n')

df3['avg'] = df3[['math', 'physics', 'history']].mean(axis=1)
print(f'Әр студенттің орташа балы бар жаңа DataFrame:\n{df3}\n')

high_avg_students = df3[df3['avg'] > 85]
print(f'Орташа балы 85-тен жоғары студенттер:\n{high_avg_students}')
```
